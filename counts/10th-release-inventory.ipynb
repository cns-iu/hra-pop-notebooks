## Global variables

```python
hra_pop_version = 'v1.0'
branch = 'main'
```

## Load data

```python
url = f'https://raw.githubusercontent.com/x-atlas-consortia/hra-pop/refs/heads/{branch}/output-data/{hra_pop_version}/reports/universe-ad-hoc/sankey.csv'

df_sankey = pd.read_csv(url)
df_sankey
```

```python
# could also use https://apps.humanatlas.io/api/grlc/hra-pop.html#get-/cell-types-in-atlas
url = f"https://raw.githubusercontent.com/x-atlas-consortia/hra-pop/refs/heads/{branch}/output-data/{hra_pop_version}/reports/atlas-ad-hoc/cell-types-in-anatomical-structurescts-per-as.csv"

headers = {
    'Accept': 'text/csv'
}

data = requests.get(url=url, headers=headers).text

df_as = pd.read_csv(StringIO(data))
df_as
```

```python
crosswalk_azimuth = pd.read_csv(
    'https://cdn.humanatlas.io/digital-objects/ctann/azimuth/v1.2/assets/azimuth-crosswalk.csv', skiprows=10)
crosswalk_azimuth
```

```python
crosswalk_celltypist = pd.read_csv(
    'https://cdn.humanatlas.io/digital-objects/ctann/celltypist/v1.1/assets/celltypist-crosswalk.csv', skiprows=10)
crosswalk_celltypist
```

```python
crosswalk_popv = pd.read_csv(
    'https://cdn.humanatlas.io/digital-objects/ctann/popv/v1.2/assets/popv-crosswalk.csv', skiprows=10)
crosswalk_popv
```

```python
crosswalk_vccf = pd.read_csv(
    'https://cdn.humanatlas.io/digital-objects/ctann/vccf/v1.0/assets/vccf-crosswalk.csv', skiprows=10)
crosswalk_vccf
```

```python
# extract CL IDs and labels with tool
extract = ['Organ_ID', 'Annotation_Label',
           'Annotation_Label_ID', 'CL_ID', 'CL_Label', 'CL_Match']

# Extract the columns from each DataFrame
az_selected = crosswalk_azimuth[extract].assign(tool='azimuth')
ct_selected = crosswalk_celltypist[extract].assign(tool='celltypist')
popv_selected = crosswalk_popv[extract].assign(tool='popv')
vccf_selected = crosswalk_vccf[extract].assign(tool='vccf')

# Concatenate them into one DataFrame
df_crosswalks_combined = pd.concat(
    [az_selected, ct_selected, popv_selected, vccf_selected], ignore_index=True)

df_crosswalks_combined
```

```python
# taken from https://github.com/hubmapconsortium/hra-workflows-runner/blob/main/src/gtex/downloader.js#L24-L48
ORGAN_MAPPING = {
    "UBERON:0001255": "bladder",
    "UBERON:0000178": "blood",
    "UBERON:0002371": "bone_marrow",
    "UBERON:0000970": "eye",
    "UBERON:0000948": "heart",
    "UBERON:0000059": "large_intestine",
    "UBERON:0002107": "liver",
    "UBERON:0002048": "lung",
    "UBERON:0000029": "lymph_node",  # original note: could be UBERON:0002509?
    "UBERON:0001911": "breast",      # 'mammary' also maps to this ID
    "UBERON:0001264": "pancreas",
    "UBERON:0002367": "prostate",
    "UBERON:0002097": "skin",
    "UBERON:0002108": "small_intestine",
    "UBERON:0002106": "spleen",
    "UBERON:0002370": "thymus",
    "UBERON:0003126": "trachea",
    "UBERON:0000995": "uterus",
    "UBERON:0004537": "vasculature",
    "UBERON:0002469": "esophagus mucosa",
    "UBERON:0004648": "esophagus muscularis",
    "UBERON:0001134": "skeletal muscle",
    "UBERON:0002113": "kidney",  # manually added
    "UBERON:0001347": "white adipose tissue",  # manually added
    "UBERON:0002373": "palatine tonsil",  # manually added
    "UBERON:0001013": "adipose tissue",  # manually added
    "UBERON:0001384": "primary motor cortex",  # manually added
    "UBERON:0002421": "hippocampal formation",  # manually added
    "UBERON:0000006": "islets of Langerhans",    # manually added
    "UBERON:0000160": "intestine",                    # manually added
    "UBERON:0000079": "male reproductive organ",       # manually added
    "UBERON:0001004": "respiratory system",             # manually added
    "UBERON:0001723": "tongue",                         # manually added
    "UBERON:0002509": "mesenteric lymph node",           # manually added
    "UBERON:0000167": "oral cavity",                    # manually added
    "UBERON:0001043": "esophagus",                      # manually added
    "UBERON:0002372": "tonsil",  # manually added
    "UBERON:0002450": "decidua"  # manually added


}


def try_get_organ_label(id: str):
  try:
    result = ORGAN_MAPPING[id]
    return result
  except:
    print(f'no match for {id}')
    return id
```

## HRApop dataset counts from Sankey data

```python
organ_counts = (
    df_sankey[df_sankey['is_atlas_dataset'] == True].drop_duplicates(['unique_dataset_id'])['organ_name']
    .value_counts()
    .reset_index()
)

organ_counts.columns = ['organ_name', 'count']
organ_counts
```

## 3D AS counts

```python
as_counts = (
    df_as.drop_duplicates(['as_label'])['organ']
    .value_counts()
    .reset_index()
)

as_counts.columns = ['organ', 'count']
as_counts
```

## CTann counts from crosswalks

```python
pivot_df = (
    df_crosswalks_combined.groupby('tool')['Organ_ID']
    .agg(lambda x: sorted(set(try_get_organ_label(str(v)) for v in x.dropna())))
    .reset_index()
)

long_df = pivot_df.explode('Organ_ID').reset_index(drop=True)

long_df
```
