## Get all HuBMAP IDs in the HRApop Atlas

```python
# install and import libraries
%pip install pandas requests

import pandas as pd
import requests 
import json
```

```python
# global settings
hra_pop_version = 'v1.0'
branch = 'main'
```

```python
# load HRApop reports
url = f'https://raw.githubusercontent.com/x-atlas-consortia/hra-pop/refs/heads/{branch}/output-data/{hra_pop_version}/reports/universe-ad-hoc/sankey.csv'
sankey = pd.read_csv(url)
sankey
```

```python
sankey[(sankey['dataset_id'].str.contains('hubmap')) & (sankey['is_atlas_dataset'] == True)]['dataset_id'].nunique()
```

```python
sankey[(sankey['portal'].str.contains('HuBMAP')) & (
    sankey['is_atlas_dataset'] == True)]['dataset_id'].nunique()
```

```python
# filter for HuBMAP dataset IDs
hubmap_ids = sankey[(sankey['portal'].str.contains('HuBMAP')) & (sankey['is_atlas_dataset'] == True)][[
    'portal', 'dataset_id', 'organ_name']]

hubmap_ids_filtered = hubmap_ids.drop_duplicates(subset=['dataset_id'], keep='first')
hubmap_ids_filtered
```

```python
# isolate UUID
hubmap_ids_filtered['uuid'] = hubmap_ids_filtered['dataset_id'].apply(
    lambda d: d.split('/')[len(d.split('/'))-1])
hubmap_ids_filtered
```

```python
# add HuBMAP IDs in addition to UUIDs
entity_api = f'https://entity.api.hubmapconsortium.org/datasets/'

def try_hubmap_id(url):
  """Tries to get a hubmap_id from an API response

  Args:
      url (str): URL for Entity API 
  """
  response = ''
  try:
    response = json.loads(requests.get(url).text)['hubmap_id']
    print(response)
    return response
  except:
    response = "no hubmap_id found"
    print(response)
    return response
    

hubmap_ids_filtered['hubmap'] = hubmap_ids_filtered['dataset_id'].apply(lambda d: try_hubmap_id(d))
hubmap_ids_filtered
```

```python
# adjust column names
hubmap_ids_filtered.rename(columns={'hubmap': 'hubmap_id'}, inplace=True)
hubmap_ids_filtered_export = hubmap_ids_filtered[hubmap_ids_filtered['hubmap_id'].str.contains('HBM')]
hubmap_ids_filtered_export
```

```python
# export
hubmap_ids_filtered_export.to_csv(
    'output/hubmap_ids_hra-pop_atlas.csv', index=False)
```
