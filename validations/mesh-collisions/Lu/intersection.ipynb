```python
total_df['source_extraction_sites_number'] = total_df['source'].apply(lambda x: collision_counter[x])
```

```python
total_df['target_extraction_sites_number'] = total_df['target'].apply(lambda x: collision_counter[x])
```

```python
total_df['sum_extraction_sites_number'] = total_df.apply(lambda x: len(x['source_extraction_sites_number'] & x['target_extraction_sites_number']), axis=1)
```

```python
total_df['common_extraction_sites'] = total_df.apply(lambda x: x['source_extraction_sites_number'] & x['target_extraction_sites_number'], axis=1)
```

```python
total_df = total_df.sort_values(by='sum_extraction_sites_number', ascending=False)
```

```python
total_df
```

```python
intersection_space_collision_api = 'http://localhost:8080/get-collisions-intersection-space'
```

```python
exact_collisions_with_intersection_space = []

for index, row in total_df.iterrows():
    count = 0
    for iri in row['common_extraction_sites']:
        # get extraction site
        extraction_site = ""
        response = requests.get(base_url_extraction_site+iri)

        if response:
            extraction_site = json.loads(response.text)
            extraction_site['pair1'] = row['source']
            extraction_site['pair2'] = row['target']
        else:
            raise Exception(f"Non-success status code: {response.status_code}")

        # send extraction site to /collisions endpoints
        r = requests.post(url=intersection_space_collision_api, json=extraction_site)
        r = r.json()
        if 'result' in r and r['result'] == 1:
            count += 1

    exact_collisions_with_intersection_space.append(count)

total_df['exact_collisions_with_intersection_space'] = exact_collisions_with_intersection_space
```

```python
filtered_df = total_df[total_df['exact_collisions_with_intersection_space'] > 0]
```
