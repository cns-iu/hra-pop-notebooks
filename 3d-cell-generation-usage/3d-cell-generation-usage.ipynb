## Exemplary call to visualize cell type population for the renal pyramid A in the male, left kidney

```python
# get ASpop for the male, left kidney
response = requests.get('https://apps.humanatlas.io/api/grlc/hra-pop/cell_types_in_anatomical_structurescts_per_as.csv', headers = {'accept':'text/csv'})
df_as_pop = pd.read_csv(StringIO(response.text))
df_as_pop
```

```python
# subset to only keep male left kidney, renal pyramid, azimuth
df_renal_pyramid = df_as_pop[(df_as_pop['organ'] == 'Left kidney')& (df_as_pop['as_label'] == 'renal pyramid') & (df_as_pop['tool'] == 'azimuth') & (df_as_pop['sex']=='Male')]
df_renal_pyramid
```

```python
# filter out cells below threshold such that only n_max CTs remain
n_max = 10

top10 = df_renal_pyramid.nlargest(n_max, "cell_percentage")
top10
```

```python
# transform into dict so it can be used in the payload for the API request
```

```python
node_distribution = dict(zip(top10['cell_label'], top10['cell_percentage']))
pprint(node_distribution)
```

```python
def call_3d_cell_api(file: str, scene_node: str, num_nodes: int, node_distribution: dict):
  """Wrapper function to make a call to the 3D Cell Generation API using a payload constructed with user-provided arguments

  Args:
      file (str): the link to the GLB file of the organ, see, e.g., https://apps.humanatlas.io/api#get-/v1/reference-organs
      scene_node (str): the scene node of the AS
      num_nodes (int): the max. total number of cells
      node_distribution (dict): a dict with key-value-pairs for cell label and percentages [0,1]
  """

  url = 'https://apps.humanatlas.io/api/v1/mesh-3d-cell-population'

  payload = {
      'file': file,
      'file_subpath': scene_node,
      'num_nodes': num_nodes,
      'node_distribution': node_distribution
  }

  headers = {
      'Content-Type': 'application/json'
  }

  try:
      response = requests.post(url, data=json.dumps(payload), headers=headers)
      response.raise_for_status()  # Raise an exception for HTTP errors (4xx or 5xx)
      return pd.read_csv(StringIO(response.text))
      print("Response Status Code:", response.status_code)
      print("Response JSON:", response.json())
  except requests.exceptions.RequestException as e:
      print(f"An error occurred: {e}")
      return None
```

```python
# make request
result = call_3d_cell_api(
    file='https://cdn.humanatlas.io/digital-objects/ref-organ/kidney-male-left/v1.3/assets/3d-vh-m-kidney-l.glb',
    scene_node='VH_M_renal_pyramid_L_a',
    num_nodes=10000,
    node_distribution=node_distribution
)
result
```

```python
# Next, let's define a function that turns a DataFrame into a node list that can then be passed into the CdeVisualization or NodeDistVis widget
def make_node_list(df: pd.DataFrame, is_3d: bool = False):
  """Turn a DataFrame into a list of dicts for passing them into a HRA widget

  Args:
      df (pd.DataFrame): A DataFrame with cells
  """

  # If the df does not have a z-axis column, let's add one and set all cells to 0
  if not is_3d:
    df.loc[:, ('z')] = 0

  node_list = [{'x': row['x'], 'y': row['y'], 'z': row['z'], 'Cell Type': row['Cell Type']}
               for index, row in df.iterrows()]

  return node_list
```

```python
node_list = make_node_list(result, True)
node_list
```

## Create and use `node-dist-vis` widget to display cells in 3D

```python
node_dist_vis = NodeDistVis(
    nodes=node_list,
    node_target_key='Cell Type',
    node_target_value='',
    max_edge_distance=1000
)
display(node_dist_vis)
```
