## A comparison of genes between MOSDAP and HRApop/ASCT+B tables

Question: Can I presume, or can you tell me if there are any MOSDAP genes that only appear in one tissue in the other references or in no ASCT+B tables/references?

## Install and import libraries

```python
%pip install requests pandas seaborn

import requests
import pandas as pd
import seaborn as sns
from io import StringIO
from pprint import pprint
```

## Load data

```python
hra = pd.read_csv('data/MOSDAP HRA Gene Evidence - HRA ASCT+B.csv')
hra['source'] = 'hra'
hra = hra.rename(columns={'gene': 'hgnc'})
hra
```

```python
hrapop = pd.read_csv('data/MOSDAP HRA Gene Evidence - HRApop Datasets.csv')
hrapop['source'] = 'hrapop'
hrapop = hrapop.rename(columns={'gene': 'hgnc'})
hrapop
```

```python
kpmp = pd.read_csv('data/MOSDAP HRA Gene Evidence - HRA KPMP-HuBMAP (Kidney Only).csv')
kpmp['source'] = 'kpmp'
kpmp = kpmp.rename(columns={'gene': 'hgnc'})
kpmp
```

```python
mosdap = pd.read_csv('data/MOSDAP ST Gene Lists.xlsx - MOSDAP_480_list.csv')
mosdap = mosdap.rename(
  columns={
    'Gene':'gene_label'
  }
)
mosdap
```

```python
url = 'https://apps.humanatlas.io/api/grlc/hra/ensembl-lookup.csv'

# Get the response
response = requests.get(url)
response.raise_for_status()  # will raise an error if request fails

# Wrap response text in StringIO so pandas can read it like a file
json_data = StringIO(response.text)

# Load into DataFrame
look_up = pd.read_csv(json_data)
look_up = look_up.rename(columns={'gene_name': 'gene_label'})
look_up
```

## Preprocess

### HRA (add organ IRIs)

```python
df_combined = pd.concat([hra, hrapop, kpmp])
df_combined
```

```python
# add ref-organ
url = 'https://apps.humanatlas.io/api/v1/reference-organs'

# Get the response
response = requests.get(url)
response.raise_for_status()  # will raise an error if request fails

json_data = response.json()
ref_organs = json_data

uberon_ids = set([organ['representation_of']for organ in ref_organs])
  
```

```python
# make look-up between as_id and ref_organ
def get_list_as(organ_id:str):
  """Gets a list of all AS in the provided organ

  Args:
      organ_id (str): Organ ID
  """
  url = f'https://apps.humanatlas.io/api/grlc/hra/as-parts?location='+organ_id
  
  headers = {
    'accept':'text/csv'
  }
  
    # Get the response
  response = requests.get(url, headers=headers)
  response.raise_for_status()  # will raise an error if request fails

  # Wrap response text in StringIO so pandas can read it like a file
  csv_data = StringIO(response.text)

  # Load into DataFrame
  df = pd.read_csv(csv_data)
  return df
```

```python
df_organ_to_as = pd.DataFrame(columns=['organ_iri', 'part_iri', 'part_label'])

for organ_id in uberon_ids:
    as_list = get_list_as(organ_id)

    # add the organ_id as a new column so we can merge cleanly
    as_list = as_list.rename(
        columns={"as_id": "part_iri", "ref_organ": "part_label"})
    as_list["organ_iri"] = organ_id

    # append all rows at once
    df_organ_to_as = pd.concat([df_organ_to_as, as_list], ignore_index=True)
    
df_organ_to_as = df_organ_to_as.rename(columns={'part_iri': 'as'})
df_organ_to_as
```

```python
df_combined_with_organ = pd.merge(df_combined, df_organ_to_as, on='as', how='inner')
df_combined_with_organ
```

### MOSDAP (add HGNC IDs)

```python
df_mosdap = pd.merge(mosdap, look_up, on='gene_label',
                     how='left').drop(columns=['ensemble'])
df_mosdap
```

## Analyze

```python
# mosdap genes that appear in no ASCT+B tables
no_asctb = df_mosdap[~df_mosdap['gene_label'].isin(df_combined_with_organ.loc[df_combined_with_organ['source'] == 'asctb', 'gene_label'])]
no_asctb
```

```python
# not in hra-pop
not_in_hrapop = df_mosdap[~df_mosdap['gene_label'].isin(
    df_combined_with_organ.loc[df_combined_with_organ['source'] == 'hrapop', 'gene_label'])]
not_in_hrapop
```

```python
# not in kpmp
not_in_kpmp = df_mosdap[~df_mosdap['gene_label'].isin(
    df_combined_with_organ.loc[df_combined_with_organ['source'] == 'kpmp', 'gene_label'])]
not_in_kpmp
```

```python
# Count distinct as_label per gene_label in df_combined
counts = (
    df_combined_with_organ.groupby("gene_label")["organ_iri"]
    .nunique()                      # number of unique as_label per gene_label
    .reset_index(name="organ_count")
)

# Get the list of unique organs per gene_label
organ_lists = (
    df_combined_with_organ.groupby("gene_label")["organ_iri"]
    .unique()  # returns a numpy array of unique organ_iri per gene_label
    .reset_index(name="organ_list")
)

# Merge with your counts if needed
counts_with_list = counts.merge(organ_lists, on="gene_label", how="left")
counts_with_list
```

```python
# Merge back onto df_mosdap by matching gene_name to gene_label
df_with_counts = df_mosdap.merge(
    counts_with_list, left_on="gene_label", right_on="gene_label", how="left")

# Fill NaN (if a gene_name never appears)
df_with_counts["organ_count"] = df_with_counts["organ_count"].fillna(0).astype(int)
df_with_counts
```

```python
only_one = df_with_counts[df_with_counts['organ_count'] == 1]
only_one
```

## Unique genes by source

```python
# step 1: drop duplicates on (gene, source)
unique_pairs = df_combined[['gene_label', 'source']
                           ].drop_duplicates(subset=["gene_label", "source"])

# step 2: count per gene
result = (
    unique_pairs.groupby("gene_label")
    .size()
    .reset_index(name="unique_source_count")
)

above_two = result[result['unique_source_count'] > 1]
above_two
```

```python
# step 1: drop duplicates on (gene, source)
unique_pairs = df_not_in_mosdap[['gene_label', 'source']
                           ].drop_duplicates(subset=["gene_label", "source"])

# step 2: count per gene
result = (
    unique_pairs.groupby("gene_label")
    .size()
    .reset_index(name="unique_source_count")
)

above_two = result[result['unique_source_count'] > 1]
above_two
```

## Gene by source

```python
# create df with all sources
df_mosdap['source'] = 'mosdap'
df_not_in_mosdap = df_not_in_mosdap.rename(columns={
  'gene_name':'gene_label'
})
df_not_in_mosdap
```

```python
df_combined
```

```python
df_all_genes_with_source = pd.concat([df_mosdap, df_not_in_mosdap, df_combined])
df_all_genes_with_source
```

```python
df_all_genes_with_source['gene_label'].nunique()
```

```python
df_unique = df_all_genes_with_source[[
    'gene_label', 'source']].drop_duplicates()
df_unique
```

```python
# assuming df has columns: gene_label, source
df_wide = (
    df_unique.assign(val=1)  # add helper column
    .pivot_table(index="gene_label",
                 columns="source",
                 values="val",
                 aggfunc="max",
                 fill_value=0)
    .reset_index()
)

# optionally rename columns to match your desired style
df_wide = df_wide.rename(columns={
    'mosdap': 'in_mosdap',
    'hra': 'in_hra',
    'hrapop': 'in_hra_pop',
    'kpmp': 'in_kpmp'
})
df_wide
```
